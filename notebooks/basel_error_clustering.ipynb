```python
import numpy as np
import trimesh

from utils.curvature_calculation import calculate_curvature_maps
from utils.spectral_clustering import calculate_spectral_decomposition, calculate_optimal_n_clusters, cluster_curvature_maps
from utils.spectral_clustering import get_age_error_correlation_statistics, get_ethnicity_error_correlation_statistics
```

## Basel 3DMM: cluster curvature-based error maps @nose

```python
face_region = 'nose'

gt_curvature_maps, rec_curvature_maps, diff_curvature_maps = calculate_curvature_maps(face_region=face_region)
diff_curvature_maps = np.load(f'data/results/curvature_maps/{face_region}/{face_region}_curvature_maps_diff.npy')
spectral_coefs, evecs = calculate_spectral_decomposition(diff_curvature_maps, face_region, n_eig=512)
get_age_error_correlation_statistics(diff_curvature_maps, evecs, spectral_coefs, face_region)
calculate_optimal_n_clusters(spectral_coefs)
```

```python
n_clusters = 9
cluster_labels = cluster_curvature_maps(n_clusters, diff_curvature_maps, spectral_coefs, evecs, face_region)
get_ethnicity_error_correlation_statistics(cluster_labels, face_region)
```
